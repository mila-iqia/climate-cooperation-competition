```python
#!/usr/bin/env python3
"""
Example script demonstrating how to use the create_plots and compute_consumption_breakdown functions.
The latter now also includes a heatmap of inter‑regional imports as % of recipient consumption
and overlays export ratios as a line on each region’s bar chart.
"""

import os
from rice_jax.utils import create_plots, compute_consumption_breakdown


episode_logs_dir = "episode_logs"
# NOTE: Before running this cell, some logs have to be generated in the episode_logs_dir

assert os.path.exists(episode_logs_dir), "Episode logs directory not found"

# Find the most recent JSON log file
json_files = [f for f in os.listdir(episode_logs_dir) if f.endswith(".json")]
assert len(json_files) > 0, "No JSON files found in '{episode_logs_dir}'."

# Use the most recent file (assuming timestamp-based naming)
latest_file = sorted(json_files)[-1]
json_path = os.path.join(episode_logs_dir, latest_file)

print(f"Generating plots from: {json_path}")

# Create plots with default parameters (now creates a single combined plot)
plot_files = create_plots(
    json_log_path=json_path, 
    output_dir="plots", 
    parameter_keys=[
        "global_temperature",  # Combined temperature plot
        "production_all_regions",
        "utility_all_regions",
        "actions.savings_rate",
        "actions.mitigation_rate",
        "gross_output_all_regions",
        "damages_all_regions",
    ],
    figsize=(12, 8), 
    dpi=300
)

print(f"Created {len(plot_files)} plot:")
for plot_file in plot_files:
    print(f"  - {plot_file}")
```

```python

# Generate consumption breakdown visualization
breakdown_file = compute_consumption_breakdown(
    json_log_path=json_path,
    output_dir="plots",
    figsize=(14, 8),
    dpi=300
)

print(f"Created consumption breakdown plot:")
print(f"  - {breakdown_file}")
```
